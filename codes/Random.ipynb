```python
from collections import *
```

```python
input_file = "inputs/random.txt"
word_count = defaultdict(int)
linecount = 0
max_count = 0
max_words = list()
with open (input_file,'r') as file:
    for line in file:
        words = line.split()
        for word in words:
            word_count[word.lower()]+=1
                     
        linecount = linecount + 1
#print(word_count)

for word in word_count:
    
    count = word_count[word]
    if count > max_count:
        max_count = count
counts = [value for key, value in word_count.items()]
counts.sort(reverse=True)
print(counts)
for count in counts:
    wordList = [key for key, value in word_count.items() if value == count]
    for i in range(len(wordList)):
        print(wordList[i]+":",word_count[wordList[i]],":",count)
```
