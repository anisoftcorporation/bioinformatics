```python
def MaximumSkew(pattern):
    pattern_len = len(pattern)
    max_skew_pos = list()
    max_skew_val = -99
    skew = list()
    skew.insert(0,0)
    index = 0
    while index < pattern_len:
      
      cur = pattern[index]
      skew_index = index+1
      
      
      if cur == 'C':
            skew.insert(skew_index,skew[index]-1)
            
      elif cur == 'G':
            skew.insert(skew_index,skew[index]+1)
            
      else:
            skew.insert(skew_index,skew[index])
      if skew[skew_index] == max_skew_val:
             max_skew_pos.append(skew_index)
      elif skew[skew_index] > max_skew_val:
            max_skew_val = skew[skew_index]
            max_skew_pos = list()
            max_skew_pos.append(skew_index)
      index = index+1
    result = " ".join(map(str,max_skew_pos))
    print(skew)
    return result

print(MaximumSkew("GCATACACTTCCCAGTAGGTACTG"))
```

```python
def Motif_Score(motif):
    final_score = 0 
    k = len(motif[0])
    t = len(motif)
    for j in range(k):
        column_score_map = defaultdict(int)
        column_score = 0 
        for i in range(t):
            column_score_map[motif[i][j]]+=1
        
        max_value_key = max(column_score_map, key=column_score_map.get)
        
       
        
        for key in column_score_map:
           
            if key != max_value_key:
                   
                    column_score += column_score_map[key]
        final_score += column_score
    return final_score
def Count_Matrix(motif):
    k = len(motif[0])
    t = len(motif)
    count_matrix = []
    A_COUNT = list()
    C_COUNT = list()
    G_COUNT = list()
    T_COUNT = list()
    
    for j in range(k):
        column_score_map = defaultdict(int)
        column_score = 0 
        for i in range(t):
            column_score_map[motif[i][j]]+=1
            
        A_COUNT.append(column_score_map['A'])
        C_COUNT.append(column_score_map['C'])
        G_COUNT.append(column_score_map['G'])
        T_COUNT.append(column_score_map['T'])
    
    count_matrix.append(A_COUNT)
    count_matrix.append(C_COUNT)
    count_matrix.append(G_COUNT)
    count_matrix.append(T_COUNT)
    return count_matrix
def Profile_Matrix(count_matrix):
    profile_matrix = []
    k = len(count_matrix[0])
    column_sums = [0]*k
    for row in count_matrix:
        for i in range(k):
            column_sums[i] += row[i]
            
    print(column_sums)
    for row in count_matrix:
        prob = [0]*k
        for i in range(k):
            prob[i] = row[i]/column_sums[i]
        profile_matrix.append(prob)
    return profile_matrix
    

DNA = "GGCGTTCAGGCA AAGAATCAGTCA CAAGGAGTTCGC CACGTCAATCAC CAATAATATTCG"
k=3
t=5
score = float('inf')
motifs = []
DNA_List = DNA.split()
for i in range(len(DNA_List)):
    text = DNA_List[i]
    row = [] 
    for j in range(k):
        row.append(text[j])
    motifs.append(row)   
    
    
def Get_Profile_Matrix(motifs):
    count_matrix = Count_Matrix(motifs)
    profile_matrix = Profile_Matrix(count_matrix)
    return profile_matrix

for motif in motifs:
    print(motif)
final_score = Motif_Score(motifs) 


profile_matrix = Get_Profile_Matrix(motifs)
for row in profile_matrix:
    print(row)
```
