```python
def MaximumSkew(pattern):
    pattern_len = len(pattern)
    max_skew_pos = list()
    max_skew_val = -99
    skew = list()
    skew.insert(0,0)
    index = 0
    while index < pattern_len:
      
      cur = pattern[index]
      skew_index = index+1
      
      
      if cur == 'C':
            skew.insert(skew_index,skew[index]-1)
            
      elif cur == 'G':
            skew.insert(skew_index,skew[index]+1)
            
      else:
            skew.insert(skew_index,skew[index])
      if skew[skew_index] == max_skew_val:
             max_skew_pos.append(skew_index)
      elif skew[skew_index] > max_skew_val:
            max_skew_val = skew[skew_index]
            max_skew_pos = list()
            max_skew_pos.append(skew_index)
      index = index+1
    result = " ".join(map(str,max_skew_pos))
    print(skew)
    return result

print(MaximumSkew("GCATACACTTCCCAGTAGGTACTG"))
```

**Randomized Motiff Search**
Now we will be using **Monte-Carlo** simulation method in our Algorithm. As per this, we will randomly select K-mer from each of the DNA string, and use that as the starting motiff to create a starting profile.

After that we will be Iterating over all the DNA strings and find Most probable K-mer from each and with them after each iteration, we will have a new Motiff.

We will calculate score and until we get a better score, we will keep doing the same loop. in next loop, this motiff will be used as base for starting Profile.




```python


DNA = "GCACATCATTAAACGATTCGCCGCATTGCCTCGATTAACC TCATAACTGACACCTGCTCTGGCACCGCTCATCCAAGGCC AAGCGGGTATAGCCAGATAGTGCCAATAATTTCCTTAACC AGTCGGTGGTGAAGTGTGGGTTATGGGGAAAGGCAAGGCC AACCGGACGGCAACTACGGTTACAACGCAGCAAGTTAACC AGGCGTCTGTTGTTGCTAACACCGTTAAGCGACGAAGGCC AAGCTTCCAACATCGTCTTGGCATCTCGGTGTGTTTAACC AATTGAACATCTTACTCTTTTCGCTTTCAAAAAAAAGGCC"
k = 6
t= 8
all_best_motifs = []
BEST_SCORE = float('inf')
for _ in range(1000):
    best_motifs = RandomizedMotifSearch(DNA, k, t)
    if Motif_Score(best_motifs) < BEST_SCORE:
        BEST_SCORE = Motif_Score(best_motifs)
        all_best_motifs = best_motifs

result = []
# Print the best motifs space-separated
for motiff in all_best_motifs:
    result.append("".join(motiff))
final_string  = " ".join(result)
print(final_string)
```
